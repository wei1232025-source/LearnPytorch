```python
import torch
from torch.utils.data import TensorDataset

# 创建一个有100个特征的数据集, 服从标准正态分布
X = torch.randn(100, 2)

Y = torch.randint(0, 2, (100, ))

dataset = TensorDataset(X, Y)
```

```python
print(X)
print(Y)
print(len(dataset))
```

```python
x1, y1 = dataset[0]
print(x1, y1)
```

```python
# DIY dataset
from torch.utils.data import Dataset

class Mydataset(Dataset):

    def __init__(self, X, Y):
        self.X = X
        self.Y = Y

    def __getitem__(self, index):
        return X[index], Y[index]

    def __len__(self):
        return len(self.X)
```

```python
# Dataloader：实现自动分批次训练

from torch.utils.data import DataLoader

loader = DataLoader(
    dataset,
    batch_size=10,
    shuffle=True
)

for x, y in loader:
    print(x.shape)
    print(y.shape)
```

```python
# 划分训练集和验证集
from torch.utils.data import random_split

train_dataset, val_dataset = random_split(
    dataset,
    [80, 20]
)
```
