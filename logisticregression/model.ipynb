```python
import torch

x = torch.tensor([
    [0.],
    [1.],
    [2.],
    [4.],
    [5.]
                  ])
y = torch.tensor([
    [0.],
    [0.],
    [0.],
    [1.],
    [1.]
                  ])


w = torch.tensor([[0.0]],
                 requires_grad=True)
b = torch.tensor([0.0],
                 requires_grad=True)
```

```python
epochs = 1000
lr = 0.05

for epoch in range(epochs):

    z = x@w + b

    p = torch.sigmoid(z)

    L = -(y*torch.log(p) + (1-y)*torch.log(1-p))

    L.backward()
    w -= lr*w.grad
    b -= lr*b.grad

    if (epoch+1)%100 == 0:
        print(f"[{epoch+1}/{epochs}]-----loss : {L}")
```
